```python
import numpy as np
import pandas as pd
```

**Finding Missing Data**

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

**Removing Missing Data: Works on Rows**

```python
df
```

```python
df.dropna()  #works on rows: Will drop those rows having null values
```

```python
df
```

```python
df.dropna(thresh=3)  #kam se kam 3 non null values ho to drop ni hogi baaki drop
```

**Filling the Missing Data**

```python
df
```

```python
df.fillna(0)
```

```python
values={'A':1000,'B':2000,'C':3000,'D':4000}
```

```python
df.fillna(value=values)
```

```python
df.fillna(df.mean())
```
