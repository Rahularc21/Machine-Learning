```python
#make a new column for episode count
#title se episode count niklagee
df.loc[0]['Title']
```

```python
#extract 64 eps
def extractEpisodes(txt):
  data=''
  check=False
  for i in txt:
    if i==')':
      break
    if i=='(':
      check=True
      continue
    if(check==True):
      data += i
  return data
```

```python
extractEpisodes(df.loc[0]['Title'])
```

```python
df['Title'].apply(extractEpisodes).head(2)   #feature extraction
```

```python
df['Episodes']=df['Title'].apply(extractEpisodes)
```

```python
df.head(2)
```

```python
df['Episodes']=df['Episodes'].str.replace(" eps","")  #extracting only count of episodes
```

```python
df.head(2)
```

```python
df.loc[0]['Episodes']   #it is in string
```

```python
df['Episodes']=df['Episodes'].astype(int)   #converting in int
```

```python
#make a new column for time stamp
```

```python
df.loc[0]['Title']
```

```python
# Apr 2009 - Jul 2010 extract this 18 letters after the bracket
```

```python
def extractTimeStamp(txt):
  check=False
  data=""
  for i in range(len(txt)):
      if txt[i]==')':
        for j in range(i+1,i+20):
          data+=txt[j]
  return data
```

```python
df['Title'].apply(extractTimeStamp).head(2)
```

```python
df['TotalTime']=df['Title'].apply(extractTimeStamp)
```

```python
df.head()
```

```python
#lets extract the total months from total time

def total_months(x):
    start, end = x.split(' - ')
    start = pd.to_datetime(start, format='%b %Y')
    end = pd.to_datetime(end, format='%b %Y')
    return (end.year - start.year) * 12 + (end.month - start.month) + 1

df['Months'] = df['TotalTime'].apply(total_months)
```

```python
#which anime has the highest score
#give me top 5 highest scoring anime
#which anime has the highest episode count
#animes with top 5 episode count
#which is the longest running anime
```

```python
#which is the longest running anime
df[df['Months']==df['Months'].max()]
```
