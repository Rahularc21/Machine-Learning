**Series:**

A Series is a one-dimensional labeled array capable of **holding any data type**.The axis labels are collectively called the index.


**Creating series: Ye vertically lgte haiii**

```python
import numpy as np
import pandas as pd
```

```python
labels=['a','b','c']
my_list=[10,20,30]
arr=np.array([10,20,30])
d={'a':10,2:20,3:30}
```

```python
#creating series now
```

```python
pd.Series(my_list)
```

0 1 2 are labels or indices and 10 20 30 are data points..

```python
pd.Series(my_list,index=labels)  #custom labels
```

```python
pd.Series(arr)
```

```python
pd.Series(d)  #by default labels are keys
```
