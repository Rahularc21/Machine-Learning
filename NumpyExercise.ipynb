**Valid Suduko**

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
valid=True
if (np.sum(s,axis=1)!=45).any():     #axis use kiya to ek poora array dega na true false ka to
  print("Not a Valid suduko")        #any us poore array pe check krega ki any value !=45
if(np.sum(s,axis=0)!=45).any():
  print("Not a Valid suduko")
```

```python
# ab 3X3 mein check kregeee
```

```python
s[0:3,0:3]    # ese hi sab ka sum 45 hona chahiyee
```

```python
for i in range(0,9,3):
  for j in range(0,9,3):
    if((np.sum(s[i:i+3,j:j+3]))!=45):     #ye ek value dega to siddha check (no any used or all)
      print("Not a valid suduko")
      valid=False
      break;
if(valid):
  print("Valid")
```
